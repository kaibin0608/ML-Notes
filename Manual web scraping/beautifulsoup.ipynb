# Things need to know 

🧠 A quick checklist before you scrape
1. ✅ Check site structure

Use your browser’s Inspect Element → find the right HTML nodes like headlines, timestamps, content blocks.

2. ⚖️ Respect terms

The disclaimer says no unauthorized copying, adaptation or reuse of content. 
Enanyang
- If scraping for commercial redistribution or public publishing, get written permission.

3. 🚫 Robots.txt

Check https://www.enanyang.my/robots.txt — if disallowed, don’t scrape those paths.

4. 🛡 Be polite

Add delays (time.sleep()) and don’t hammer the server. Consider caching if scraping repeatedly.


🏁 When sites have dynamic content

If some pages load articles via JavaScript, static scraping with requests won’t see them. In that case:
- Switch to Selenium or Playwright (automate a browser).
- Or monitor XHR network calls for API endpoints the site uses and call those directly.

reference: https://beautiful-soup-4.readthedocs.io/en/latest/#

When scraping with BeautifulSoup:

✔ You are navigating a tree

✔ Tags = Element Nodes
- <p> tells the browser:“This is a paragraph of text.”
- <a> → Anchor (Link):Used for clickable links.
- <h1> → Main Heading
- <div> → Container / Box : “Group things together.”
- <span> → Small Inline Container : it is used inside text
- <script> -> contains code, not normal page content.

✔ .text = Text Nodes

✔ ["href"] = Attributes

✔ .parent / .children = Tree Navigation

Examples: 
```python
<a href="/news">Read more</a>
```
- Tag → a
- Attribute → href
- Text → Read more


````python
# scraper-python.py
# To run this script, paste `python scraper-python.py` in the terminal

import requests
from bs4 import BeautifulSoup

````

````python
def scrape():
    
    url = 'https://www.example.com'
    response = requests.get(url)
    soup = BeautifulSoup(response.text, 'html.parser')
    title = soup.select_one('h1').text
    text = soup.select_one('p').text
    link = soup.select_one('a').get('href')

    print(title)
    print(text)
    print(link)
    print(soup.prettify())
````

````python
scrape()
````

````python
import requests
from bs4 import BeautifulSoup
from datetime import datetime
import time

# from sqlalchemy import create_engine, Column, String, Text, DateTime
# from sqlalchemy.orm import sessionmaker, declarative_base

BASE_URL = "https://www.enanyang.my"
TAG_PAGE = "/tag/ssm"  # SSM tag page

# 1) Database setup (SQLite)
# engine = create_engine("sqlite:///ssm_news.db", echo=False)
# Session = sessionmaker(bind=engine)
# Base = declarative_base()

# class Article(Base):
#     __tablename__ = "articles"
#     url = Column(String, primary_key=True)
#     title = Column(String)
#     published_date = Column(String)
#     content = Column(Text)
#     tags = Column(String)
#     scraped_at = Column(DateTime, default=datetime.utcnow)

# Base.metadata.create_all(engine)

# session = Session()

# 2) Fetch SSM tag page
headers = {
    "User-Agent": "MyScraperBot/1.0 (+https://example.com/bot-info)"
}

resp = requests.get(BASE_URL + TAG_PAGE, headers=headers)
soup = BeautifulSoup(resp.text, "html.parser")

# 3) Extract article links
links = []
for link in soup.select("a[href*='/news/']"): #*= means contains
    href = link.get("href")
    if href and "SSM" in link.text:  # basic filter
        full = BASE_URL + href
        if full not in links:
            links.append(full)

print(f"Found {len(links)} potential SSM news links.")

# 4) Crawl each article
for url in links:
    # skip if already in DB
    if session.query(Article).filter_by(url=url).first():
        print("Already saved:", url)
        continue

    print("Fetching:", url)
    art_resp = requests.get(url, headers=headers)
    art_soup = BeautifulSoup(art_resp.text, "html.parser")

    # example selectors (adjust based on actual HTML)
    title = art_soup.select_one("h1").get_text(strip=True) if art_soup.select_one("h1") else ""
    date = art_soup.select_one(".posted-on")  # adjust selector if needed
    published_date = date.get_text(strip=True) if date else ""
    body_elems = art_soup.select("div.entry-content p")
    content_text = "\n".join(p.get_text(strip=True) for p in body_elems)

    # Save to database
    article = Article(
        url=url,
        title=title,
        published_date=published_date,
        content=content_text,
        tags="SSM"
    )
    session.add(article)
    session.commit()

    print("Saved:", title)

    time.sleep(1)  # be polite
````
