```python
from firecrawl import Firecrawl
from firecrawl.v2.types import JsonFormat
from datetime import datetime
import logging
# import json 
base_url = 'https://www.sinchew.com.my/?s='


def custom_news_scraper(keywords,content_start_date,content_end_date, user_request=None):
    firecrawl = Firecrawl(api_key = api_key)

    start_dt = datetime.strptime(content_start_date, '%Y-%m-%d') if content_start_date else None
    end_dt = datetime.strptime(content_end_date, '%Y-%m-%d') if content_end_date else None
    
    validated_results = []
    all_articles = []
    
    for keyword in keywords:
        
        for url in base_url:
            website_url = f'{url}{keyword}'


            try:
                result = firecrawl.scrape(
                    url = website_url,
                    formats = [
                        JsonFormat(
                            type= 'json',
                            prompt = "Extract all dates and links from the news in the page, return dates in YYYY-MM-DD format (e.g. 2026-02-10) and map the dates to each links extracted",
                            schema = { 
                                "type" : "object",
                                "required" : [],
                                "properties" : {
                                    "results": {
                                        "type": "array",
                                            "items":{
                                                "type":"object",
                                                "required":["date","url"],
                                                "properties":{
                                                    "date":{"type":"string"},
                                                    "url":{"type":"string"}
                                                }
                                            }
                                    }
                                }
                            }
                        )
                    ]
                )

                if not result.json:
                    logger.warning(f"No JSON data returned for {website_url}")
                    continue

            except Exception as e:
                logger.error(f"Failed to scrape {website_url}: {e}")
                continue
            
            for news in result.json.get('results',[]):
                try:
                    news_date = datetime.strptime(news.get('date'), '%Y-%m-%d')
                except (ValueError, TypeError):
                    logger.warning(f"Skipping news with invalid date: {news.get('date')}")
                    continue

                if news_date and not start_dt <= news_date <= end_dt:
                    continue
                else: 
                    validated_results.append(news)

    if validated_results:
        urls = [news['url'] for news in validated_results]
        content = firecrawl.batch_scrape(
            urls,
            formats=["markdown"]
        )
        

        for item, news in zip(content.data,validated_results):
            metadata = vars(item.metadata)
            metadata['timestamp'] = news['date']
            metadata['text'] = metadata.pop('og_description','')
            all_articles.append(metadata)
        
        if user_request:
            logger.info("Running QC Layer to check article relevance...")
            all_articles = check_relevance(all_articles, user_request)
            relevant_count = sum(1 for item in all_articles if item.get('is_relevant'))
            logger.info(f"QC Result: {relevant_count}/{len(all_articles)} articles are relevant")

    else:
        logger.info(f"No news articles found for the given keyword and date range.")

    output = {"total_available": len(all_articles), "news": all_articles}
    output = {"total_available": '1000', "news": 'test'}

    output_dir = "logs"
    filename = f"news_results_{datetime.now().strftime('%Y%m%d_%H%M%S')}.json"
    filepath = save_json(output, filename, output_dir)
    
    
    # with open(filepath, 'w', encoding='utf-8') as f:
        # json.dump(output, f, indent=4, ensure_ascii=False)
    
    logger.info(f"\nAll news saved to '{filepath}'")
    return output
        
```
