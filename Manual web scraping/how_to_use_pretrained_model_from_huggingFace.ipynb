# Manual Use: Tokenizers + Models

```python
from transformers import AutoTokenizer, AutoModelForCausalLM

tokenizer = AutoTokenizer.from_pretrained("gpt2")
model = AutoModelForCausalLM.from_pretrained("gpt2")

inputs = tokenizer("Hello world", return_tensors="pt")
outputs = model.generate(**inputs, max_new_tokens=50)
print(tokenizer.decode(outputs[0], skip_special_tokens=True))
```

### Tokenizer

Tokenizer basically put a text in a mathematical representation that the model understand

```python
sequence = " Using a Transformer network is simple"
res = tokenizer(sequence)
print(res)
tokens = tokenizer.tokenize(sequence)
print(tokens)
ids = tokenizer.convert_tokens_to_ids(tokens) 
print(ids)
decoded_string = tokenizer.decode(ids)
print(decoded_string)
```

## Combine the code with pytorch and tensorflow

```python
from transformers import pipeline
from transformers import AutoTokenizer, AutoModelForSequenceClassification
import torch
import torch.nn.functional as F

model_name = "distilbert-base-uncased-finetuned-sst-2-english"
model = AutoModelForSequenceClassification.from_pretrained(model_name)

tokenizer = AutoTokenizer.from_pretrained(model_name)

classifier = pipeline("sentiment-analysis", model = model, tokenizer = tokenizer)

X_train = ["I've been waiting for a HuggingFace course my whole life.","Python is great!"]

res = classifier(X_train)
print(res)

batch = tokenizer(X_train, padding = True, truncation = True, max_length = 512, return_tensors = "pt") # "pt" means pytorch format
print(batch)

with torch.no_grad():
    outputs = model(**batch) # ** means unpack this 'batch', 
    print(outputs)
    predictions = F.softmax(outputs.logits,dim = 1) # to get the predictions
    print(predictions)
    labels = torch.argmax(predictions,dim =1) # to get the labels 
    print(labels)

```

```python
from transformers import AutoModelForSequenceClassification, AutoTokenizer

# Initialize global variables
TEST_SENTIMENT_MODEL = None
TEST_SENTIMENT_TOKENIZER = None

def load_sentiment_model():
    """Load the Malay sentiment analysis model"""
    global TEST_SENTIMENT_MODEL, TEST_SENTIMENT_TOKENIZER

    if TEST_SENTIMENT_MODEL is None or TEST_SENTIMENT_TOKENIZER is None:
        # print("Loading sentiment analysis model...")
        model_name = "Hanisnabila/Malay-Sentiment-6"
        # model_name = "mesolitica/mnli-malaysian-mistral-191M-MLM-512"
        TEST_SENTIMENT_TOKENIZER = AutoTokenizer.from_pretrained(model_name)
        TEST_SENTIMENT_MODEL = AutoModelForSequenceClassification.from_pretrained(model_name)
        # print("Sentiment analysis model loaded successfully.")

    return TEST_SENTIMENT_TOKENIZER, TEST_SENTIMENT_MODEL

# Function to analyze sentiment
def analyze_sentiment(text):
    """Analyze the sentiment of a text using the Malay sentiment model"""
    if text is None or text.strip() == "":
        return {"score": 0.0, "sentiment": "neutral"}

    try:
        # Load model if not already loaded
        tokenizer, model = load_sentiment_model()

        # Tokenize the text
        inputs = tokenizer(text, return_tensors="pt", truncation=True, max_length=512, padding=True)
        # Get model predictions
        with torch.no_grad():
            outputs = model(**inputs)
            predictions = torch.nn.functional.softmax(outputs.logits, dim=-1)
            predicted_label_id = predictions.argmax(-1).item()
            confidence_score = predictions[0][predicted_label_id].item()

        # Map the predicted label ID to sentiment
        sentiment_map = {0: "negative", 1: "neutral", 2: "positive"}
        sentiment = sentiment_map[predicted_label_id]

        return {"score": confidence_score, "sentiment": sentiment}

    except Exception as e:
        print(f"Error analyzing sentiment: {str(e)}")
        return {"score": 0.0, "sentiment": "neutral"}
```
